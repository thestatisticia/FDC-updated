## 1. Environment Check

Before running on-chain steps, make sure dependencies are installed and your `.env` is configured. The helper below validates the essentials and shows quick fixes if something is missing.


```python
import os
from pathlib import Path

REQUIRED_KEYS = [
    "PRIVATE_KEY",
    "VERIFIER_API_KEY_TESTNET",
    "FLARE_RPC_API_KEY",
]

env_path = Path(".env")
if not env_path.exists():
    print("⚠️  .env file not found. Copy .env.example and fill in your keys before continuing.")
else:
    missing = []
    with env_path.open() as fh:
        env_lines = {
            line.split("=", 1)[0].strip(): line.split("=", 1)[1].strip()
            for line in fh
            if line.strip() and not line.startswith("#") and "=" in line
        }
    for key in REQUIRED_KEYS:
        value = env_lines.get(key, "")
        if not value or value.endswith("here"):
            missing.append(key)
    if missing:
        print("⚠️  Missing values for:", ", ".join(missing))
        print("   Update .env before attempting live submissions.")
    else:
        print("✅  .env looks good. You are ready for live runs.")
```

## 2. Prepare the Attestation Request

This cell shows you the exact payload sent to the verifier service. Toggle `USE_MOCK_DATA` to inspect the sample payload without performing an HTTP call.


```python
import subprocess

USE_MOCK = True
COMMAND = ["npx", "hardhat", "run", "scripts/fdcExample/Web2Json.ts", "--network", "coston2"]
if USE_MOCK:
    COMMAND.append("--mock")

print("Running:", " ".join(COMMAND))
process = subprocess.run(COMMAND, capture_output=True, text=True)
print(process.stdout)
if process.returncode != 0:
    print("Command failed:", process.stderr)
```
